## Live Coding #1 — Compression de séquence

```python
def compress_sequence(values: list[int]) -> list[tuple[int, int]]:
    if not values:
        return []

    result = []
    current = values[0]
    count = 1

    for value in values[1:]:
        if value == current:
            count += 1
        else:
            result.append((current, count))
            current = value
            count = 1

    result.append((current, count))
    return result
        
# Test
assert compress_sequence([4, 4, 4, 2, 2, 7, 4, 4]) == [
    (4, 3), (2, 2), (7, 1), (4, 2)
]
assert compress_sequence([]) == []
assert compress_sequence([5]) == [(5, 1)]
assert compress_sequence([1, 2, 3]) == [
    (1, 1), (2, 1), (3, 1)
]
assert compress_sequence([8, 8, 8, 8]) == [(8, 4)]
```

```python
# Une solution plus compacte
from itertools import groupby


def compress_sequence(values: list[int]) -> list[tuple[int, int]]:
    return [
        (value, sum(1 for _ in group))
        for value, group in groupby(values)
    ]
```

## Live Coding #2 — Première paire de somme cible

```python
def find_pair(values: list[int], target: int) -> tuple[int, int] | None:
    seen = {}
    for j, value in enumerate(values):
        complement = target - value         

        if complement in seen:            
            return seen[complement], j

        seen[value] = j
    return None

# test
assert find_pair([2, 7, 11, 15], 9) == (0, 1)

assert find_pair([3, 2, 4], 6) == (1, 2)

assert find_pair([3, 3], 6) == (0, 1)

assert find_pair([1, 5, 3, 7, 4], 8) == (1, 2)

assert find_pair([1, 2, 3], 10) is None

assert find_pair([], 5) is None

assert find_pair([4], 8) is None

assert find_pair([-3, 4, 3, 90], 0) == (0, 2)
```

## Live Coding #3 — Parenthèses bien équilibrées

```python
def is_balanced(expression: str) -> bool:
    pairs = {
        ")": "(",
        "]": "[",
        "}": "{",
    }

    stack = []

    for char in expression:
        if char in pairs.values():
            stack.append(char)

        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False

    return not stack
# Test
assert is_balanced("") is True
assert is_balanced("hello") is True

assert is_balanced("()") is True
assert is_balanced("([]{})") is True
assert is_balanced("{[()]}") is True

assert is_balanced("(") is False
assert is_balanced(")") is False
assert is_balanced("((x)") is False

assert is_balanced("([)]") is False
assert is_balanced("{[(])}") is False

assert is_balanced("(a + b) * [c - {d / 2}]") is True
```

## Live Coding #4 — Premier caractère non répété

```python
def first_unique_char(text: str) -> str | None:
    for char in text:
        if text.count(char) == 1:
            return char
    return None
# Test
assert first_unique_char("swiss") == "w"
assert first_unique_char("aabbcddee") == "c"
assert first_unique_char("abcdef") == "a"

assert first_unique_char("aabbcc") is None
assert first_unique_char("") is None

assert first_unique_char("aA") == "a"
assert first_unique_char("x") == "x"
```

```python
# Solution avec meilleurs complexité 
def first_unique_char(text: str) -> str | None:
    counts = {}

    for char in text:
        counts[char] = counts.get(char, 0) + 1

    for char in text:
        if counts[char] == 1:
            return char

    return None
```

## Live Coding #5 — Plus longue séquence sans répétition

```python
def longest_unique_run(text: str) -> int:
    seen = set()
    left = 0
    max_length = 0

    for right, char in enumerate(text):
        while char in seen:
            seen.remove(text[left])
            left += 1

        seen.add(char)
        max_length = max(max_length, right - left + 1)

    return max_length
# Test
assert longest_unique_run("") == 0
assert longest_unique_run("a") == 1

assert longest_unique_run("abcabcbb") == 3
assert longest_unique_run("bbbbb") == 1
assert longest_unique_run("pwwkew") == 3

assert longest_unique_run("abcdef") == 6
assert longest_unique_run("abba") == 2
assert longest_unique_run("dvdf") == 3
```

## Live Coding #6 - Détection de doublon

```python
def first_duplicate(values: list[int]) -> int | None:
    seen = set()

    for value in values:
        if value in seen:
            return value
        seen.add(value)

    return None
# Test
assert first_duplicate([4, 7, 2, 7, 4]) == 7
assert first_duplicate([1, 2, 3, 1, 2]) == 1
assert first_duplicate([5, 5]) == 5

assert first_duplicate([1, 2, 3, 4]) is None
assert first_duplicate([]) is None
assert first_duplicate([8]) is None

assert first_duplicate([0, -1, 3, -1, 0]) == -1
```

## Live Coding #7 — Diviseurs d'un entier

```python
import math

def divisors(n: int) -> list[int]:
    result = []

    for divisor in range(1, math.isqrt(n) + 1):
        if n % divisor == 0:
            result.append(divisor)

            other = n // divisor
            if other != divisor:
                result.append(other)

    return sorted(result)

# Test
assert divisors(1) == [1]
assert divisors(2) == [1, 2]
assert divisors(12) == [1, 2, 3, 4, 6, 12]
assert divisors(16) == [1, 2, 4, 8, 16]
assert divisors(17) == [1, 17]
assert divisors(36) == [1, 2, 3, 4, 6, 9, 12, 18, 36]
```

## Live Coding #8 — Plus grand diviseur commun

```python
def greatest_common_divisor(a: int, b: int) -> int:
    divisors_a = set(divisors(a))

    common_divisors = [
        value
        for value in divisors(b)
        if value in divisors_a
    ]

    return max(common_divisors)
# test
assert greatest_common_divisor(18, 24) == 6
assert greatest_common_divisor(12, 8) == 4
assert greatest_common_divisor(17, 13) == 1
assert greatest_common_divisor(21, 7) == 7
assert greatest_common_divisor(100, 25) == 25
assert greatest_common_divisor(48, 18) == 6
assert greatest_common_divisor(1, 50) == 1
```

```python
# Solution plus élégante
def greatest_common_divisor(a: int, b: int) -> int:
    while b:
        a, b = b, a % b

    return a
```

## Live Coding #9 — Point le plus proche

```python
import math

def closest_to_origin(
    points: list[tuple[float, float]],
) -> tuple[float, float] | None:
    if not points:
        return None

    nearest = points[0]
    min_distance_squared = nearest[0] ** 2 + nearest[1] ** 2

    for x, y in points[1:]:
        distance_squared = x**2 + y**2

        if distance_squared < min_distance_squared:
            min_distance_squared = distance_squared
            nearest = (x, y)

    return nearest
# Test
assert closest_to_origin([]) is None

assert closest_to_origin([(3, 4)]) == (3, 4)

assert closest_to_origin([
    (3, 4),
    (1, 1),
    (5, 2)
]) == (1, 1)

assert closest_to_origin([
    (-2, -2),
    (1, 3),
    (0, 2)
]) == (0, 2)

assert closest_to_origin([
    (1, 1),
    (-1, -1),
    (4, 0)
]) == (1, 1)

assert closest_to_origin([
    (10, 10),
    (0, 0),
    (1, 1)
]) == (0, 0)
```

## Live Coding #10 — Fusion d'intervalles

```python
def merge_intervals(
    intervals: list[tuple[int, int]]
) -> list[tuple[int, int]]:
    if not intervals:
        return []
    # Sort according the first element
    sorted_intervals = sorted(intervals, key = lambda x:x[0], reverse = False)
    # assembling intervals
    result = []
    current_interval = sorted_intervals[0]
    for interval in sorted_intervals[1:]:
        if current_interval[1] < interval[0]:
            result.append(current_interval)
            # updating the current interval
            current_interval = interval
        current_interval = (current_interval[0], max(interval[1], current_interval[1]))
    # Add the last intervals
    result.append(current_interval)
    return result

# Tests
assert merge_intervals([]) == []

assert merge_intervals([(1, 3)]) == [(1, 3)]

assert merge_intervals([
    (1, 4),
    (3, 6)
]) == [(1, 6)]

assert merge_intervals([
    (1, 2),
    (2, 5)
]) == [(1, 5)]

assert merge_intervals([
    (1, 3),
    (8, 10),
    (2, 6),
    (15, 18)
]) == [
    (1, 6),
    (8, 10),
    (15, 18)
]

assert merge_intervals([
    (10, 12),
    (1, 3),
    (3, 7),
    (8, 9)
]) == [
    (1, 7),
    (8, 9),
    (10, 12)
]

assert merge_intervals([
    (1, 10),
    (2, 3),
    (4, 8)
]) == [(1, 10)]

assert merge_intervals([
    (20, 25),
    (2, 4),
    (3, 10),
    (30, 35),
    (1, 2),
    (8, 15),
    (22, 30),
    (5, 6),
    (40, 40),
    (15, 18),
]) == [
    (1, 18),
    (20, 35),
    (40, 40),
]
```
