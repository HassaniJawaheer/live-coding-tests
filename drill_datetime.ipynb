## **Exo 1**

```python
from datetime import datetime

def normalize_dates(dates: list[str]) -> list[str]:
    return [datetime.strptime(s, "%Y-%m-%d").strftime("%d/%m/%Y") for s in dates]

# tests
assert normalize_dates(["2026-02-15"]) == ["15/02/2026"]
assert normalize_dates(["2020-01-01", "1999-12-31"]) == ["01/01/2020", "31/12/1999"]

try:
    normalize_dates(["2026-02-30"])  # invalid date
    assert False, "Expected ValueError for invalid calendar date"
except ValueError:
    pass
```

## **Exo 2**

```python
from datetime import datetime, timedelta

def add_days_and_sort(stamps: list[str], days: int) -> list[str]:
    dts = []
    for s in stamps:
        dt = datetime.strptime(s, "%Y-%m-%d %H:%M")
        if dt.strftime("%Y-%m-%d %H:%M") != s:
            raise ValueError("stamp must be strictly YYYY-MM-DD HH:MM")
        dts.append(dt + timedelta(days=days))

    dts.sort()
    return [dt.strftime("%Y-%m-%d %H:%M") for dt in dts]


# Tests
assert add_days_and_sort(["2026-02-15 09:30"], 1) == ["2026-02-16 09:30"]

stamps = ["2026-02-15 09:30", "2026-02-14 10:00", "2026-02-15 09:00"]
out = add_days_and_sort(stamps, 2)
assert out == ["2026-02-16 10:00", "2026-02-17 09:00", "2026-02-17 09:30"]
assert stamps == ["2026-02-15 09:30", "2026-02-14 10:00", "2026-02-15 09:00"]

try:
    add_days_and_sort(["2026-02-15 9:30"], 1)  # not strict HH:MM
    assert False, "Expected ValueError for non-strict time format"
except ValueError:
    pass
```
