## **Exo 1**

```python
def dedup_keep_order(items: list[str]) -> list[str]:
    """Process elements while preserving order"""
    out = []
    seen = set()
    for item in items:
        if item not in seen:
            seen.add(item)
            out.append(item)
    return out

# Tests
assert dedup_keep_order(["a", "b", "a", "c", "b"]) == ["a", "b", "c"]
assert dedup_keep_order([]) == []
assert dedup_keep_order(["x", "x", "x"]) == ["x"]
assert dedup_keep_order(["A", "a", "A"]) == ["A", "a"]
```

## **Exo 2**

```python
def reconcile_access(granted: set[str], revoked: set[str], requested: set[str]) -> tuple[set[str], set[str], set[str]]:
    effective = granted | requested
    effective = effective - revoked
    newly_granted = effective - granted
    blocked = requested & revoked
    return effective, newly_granted, blocked
# Tests
granted = {"alice", "bob", "carol"}
revoked = {"bob", "eve"}
requested = {"bob", "dave", "eve", "franck"}

eff, new, blocked = reconcile_access(granted, revoked, requested)
assert eff == {"alice", "carol", "dave", "franck"}
assert new == {"dave", "franck"}
assert blocked == {"bob", "eve"}

# inputs unchanged
assert granted == {"alice", "bob", "carol"}
assert revoked == {"bob", "eve"}
assert requested == {"bob", "dave", "eve", "franck"}
```

## **Exo 3**

```python
def unpack_stats(values: tuple[int, ...]) -> tuple[int, int, list[int]]:
    if len(values) < 2:
        raise ValueError("values must contain at least 2 ints")
    first, *middle, last = values
    return first, last, middle

# tests
assert unpack_stats((10,20)) == (10,20,[])
assert unpack_stats((1,2,3,4)) == (1,4,[2,3])
assert unpack_stats((7,8,9)) == (7,9,[8])

try: 
    unpack_stats((1,))
    assert False, "Expected ValueError for len < 2"
except ValueError:
    pass
```

## **Exo 4**

```python
def rotate_head_tail(values: list[int]) -> list[int]:
    if len(values) <= 1:
        return values.copy()

    first, *middle, last = values
    first, last = last, first  # tuple swapping
    return [first] + middle + [last]

# Tests
assert rotate_head_tail([]) == []
assert rotate_head_tail([5]) == [5]
assert rotate_head_tail([1, 2]) == [2, 1]
assert rotate_head_tail([1, 2, 3, 4]) == [4, 2, 3, 1]

orig = [9, 8, 7]
out = rotate_head_tail(orig)
assert out == [7, 8, 9]
assert orig == [9, 8, 7]
```
