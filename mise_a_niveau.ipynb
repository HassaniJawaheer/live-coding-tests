```python
logs = [
    "INFO: Server started",
    "ERROR: Disk full",
    "WARNING: CPU high",
    "ERROR: Connection lost",
    "INFO: Shutdown"
]

def extract_errors(logs: list[str]) -> list[str]:
    result = []

    for log in logs:
        parts = log.split(":", maxsplit=1)

        if len(parts) == 2 and parts[0].strip() == "ERROR":
            result.append(parts[1].strip())

    return result

extract_errors(logs)
```

## **Test 2**

```python
names = ["Alice", "bob", "ALICE", "Bob", "Charlie", "alice"]

def unique_names(names: list[str]) -> list[str]:
    lower_name_form = set()
    result = []
    for name in names:
        if name.lower() not in lower_name_form:
            result.append(name)
            lower_name_form.add(name.lower())
    return result

unique_names(names)
```

```python
numbers = [3, 7, -1]

def increment(numbers: list[int]) -> list[int]:
    return [e + 1 for e in numbers]

increment(numbers)
```

## **Test 4**

```python
records = [
    {
        "first_name": "Alice",
        "last_name": "Martin",
        "age": 23,
        "passed": True,
    },
    {
        "first_name": "Bob",
        "last_name": "Smith",
        "age": 19,
        "passed": False,
    },
    {
        "first_name": "Emma",
        "last_name": "Durand",
        "age": 21,
        "passed": True,
    },
]


def format_students(records: list[dict]) -> list[str]:
    result = []

    for student in records:
        status = "PASS" if student["passed"] else "FAIL"

        result.append(
            f'{student["first_name"]} '
            f'{student["last_name"]} '
            f'({status}) - Age: {student["age"]}'
        )

    return result


print(format_students(records))
```

## **Test 5**

```python
class BankAccount:
    def __init__(self, owner: str, balance: float = 0.0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount: float) -> None:
        self.balance += amount

    def withdraw(self, amount: float) -> bool:
        if amount <= self.balance:
            self.balance -= amount
            return True
        else:
            return False

    def get_balance(self):
        return self.balance
    
    def __str__(self):
        return f"{self.owner}: {self.balance} euros"
account = BankAccount("Alice", 100.0)

print(account)

account.deposit(25)

print(account.get_balance())

print(account.withdraw(50))

print(account.get_balance())

print(account.withdraw(200))

print(account)
```
