```python
import numpy as np

# Model estimation
def fit_linear_model(
    x: np.ndarray,
    y: np.ndarray
) -> tuple[float, float]:
    # Get x dimension
    n = x.shape[0]
    
    # Matrix reformulation
    X = np.concatenate((x.reshape((n,1)), np.ones((n, 1))), axis = 1)

    # Calculate a and b, beta = np.array([a, b]) by least square method
    beta = np.linalg.inv(X.T @ X) @ X.T @ y

    return beta[0], beta[1]

# Prediction method
def predict(
    x: np.ndarray,
    a: float,
    b: float
) -> np.ndarray:
    # Get x dimension
    n = x.shape[0]
    
    # Matrix reformulation
    X = np.concatenate((x.reshape((n,1)), np.ones((n, 1))), axis = 1)
    return X @ np.array([a, b])

# MSE (Mean squared Error)
def MSE(y: np.ndarray, y_hat: np.ndarray) -> float:
    return np.mean((y - y_hat)**2)
```

```python
# Tests
import numpy as np

x = np.array([-2., -1., 0., 1., 2., 3.])
y = np.array([-3.9, -1.8, 0.1, 2.2, 3.9, 6.3])

a, b = fit_linear_model(x, y)

assert 1.8 < a < 2.2
assert -0.2 < b < 0.3

y_pred = predict(x, a, b)

assert isinstance(y_pred, np.ndarray)
assert y_pred.shape == y.shape

err = MSE(y , y_pred)

assert err >= 0
assert err < 0.1
```
