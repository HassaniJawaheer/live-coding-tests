## **Exo 1**

```python
from itertools import groupby

def runs(values: list[int]) -> list[tuple[int, int]]:
    return [(key, len(list(group))) for key, group in groupby(values)]

# Tests
assert runs([1,1,2,2,2,1]) == [(1,2),(2,3),(1,1)]
assert runs([]) == []
assert runs([7]) == [(7,1)]
assert runs([3, 3, 3]) == [(3,3)]
assert runs([1,2,1,2]) == [(1,1),(2,1),(1,1),(2,1)]
```
