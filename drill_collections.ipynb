## **Exo 1**

```python
from collections import Counter

def top_k_words(words: list[str], k: int) -> list[tuple[str, int]]:
    if k <= 0:
        return []
    return Counter(words).most_common(k)

# Tests
assert top_k_words(["a", "b", "a", "c", "b", "a"], 2) == [("a", 3), ("b", 2)]
assert top_k_words(["x", "y"], 5) == [("x", 1), ("y", 1)]
assert top_k_words(["x", "x", "y"], 0) == []
assert top_k_words([], 3) == []
```

## **Exo 2**

```python
from collections import defaultdict

def sum_by_prefix(pairs: list[tuple[str, int]]) -> dict[str, int]:
    acc = defaultdict(int)
    for key, value in pairs:
        acc[key[:2]] += value
    return dict(acc)

# Tests
pairs = [("abcd", 1), ("abZZ", 2), ("ax", 5), ("ayyy", -1), ("ax__", 3)]
assert sum_by_prefix(pairs) == {"ab":3, "ax": 8, "ay": -1}

assert sum_by_prefix([]) == {}

pairs2 = [("hi", 1), ("h", 2)]
assert sum_by_prefix(pairs2) == {"hi":  1, "h":  2}
```
