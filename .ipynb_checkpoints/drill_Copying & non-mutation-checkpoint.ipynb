## **Exo 1**

```python
def bump_nested(cfg: dict[str, dict[str, int]], key: str, subkey: str, delta: int) -> dict[str, dict[str, int]]:
    out = cfg.copy()
    if key not in out:
        raise KeyError(f"{key} does not exist in cfg")
    
    inner_dict = out[key].copy()

    if subkey not in inner_dict:
        raise KeyError(f"{subkey} does not exist in cfg[{key}]")

    inner_dict[subkey] += delta
    out[key] = inner_dict

    return out
    
# Tests
cfg = {"a": {"x": 1, "y": 2}, "b" : {"x": 10}}
out = bump_nested(cfg, "a", "x", 5)

assert out == {"a": {"x":6, "y": 2}, "b": {"x": 10}}
assert cfg == {"a": {"x": 1, "y": 2}, "b" : {"x": 10}}

# Identity checks : ony inner dict for "a" must differ
assert out is not cfg
assert out["a"] is not cfg["a"]
assert out["b"] is cfg["b"]

try: 
    bump_nested(cfg, "missing", "x", 1)
    assert False, "Expected KeyError for missing outer key"
except:
    pass

try:
    bump_nested(cfg, "a", "missing", 1)
    assert False, "Expected keyError for missing inner key"
except:
    pass
```
