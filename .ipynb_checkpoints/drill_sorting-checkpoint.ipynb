## **Exo 1**

```python
def sorted_copy_and_inplace(nums: list[int]) -> tuple[list[int], None]:
    copy_sorted = sorted(nums)
    sort_result = nums.sort()
    return copy_sorted, sort_result

# Tests
x = [3,1,2,2]
copy_sorted, sort_result = sorted_copy_and_inplace(x)
assert copy_sorted == [1,2,2,3]
assert sort_result is None
assert x == [1,2,2,3]

y = []
copy_sorted, sort_result = sorted_copy_and_inplace(y)
assert copy_sorted == []
assert y == []
```

## **Exo 2**

```python
def sort_people(records: list[tuple[str, int]]) -> list[tuple[str, int]]:
    return sorted(records, key = lambda x: (x[1], x[0]))

# tests
people = [("bob", 30),("alice", 30), ("zoe", 25), ("alice", 20)]
out = sort_people(people)

assert out == [("alice", 20), ("zoe", 25), ("alice", 30), ("bob", 30)]
assert people == [("bob", 30),("alice", 30), ("zoe", 25), ("alice", 20)]
```

## **Exo 3**

```python
def sort_tasks(tasks: list[tuple[str, int, int]]) -> list[tuple[str, int, int]]:
    # Python's sort is stable: items with equal keys keep their original relative order.
    return sorted(tasks, key=lambda t: (-t[1], t[2], t[0]))

# Tests
tasks = [
    ("email", 2, 10),
    ("build", 3, 5),
    ("deploy", 3, 5),
    ("lint", 3,7),
    ("backup", 2, 1),
]

out = sort_tasks(tasks)
assert out == [
    ("build", 3, 5),
    ("deploy", 3, 5),
    ("lint", 3, 7),
    ("backup", 2, 1),
    ("email", 2, 10),
]
assert tasks[0] == ("email", 2, 10)  # input not mutated
```

## **Exo 4**

```python
def best_candidate(cands: list[tuple[str, int, int]]) -> tuple[str, int, int]:
    if not cands:
        raise ValueError("cands cannot be empty")
    return min(cands, key=lambda t: (-t[1], t[2], t[0]))

# Tests
cands = [("bob", 10, 5), ("alice", 10, 3), ("zoe", 9, 0), ("alice", 10, 3)]
assert best_candidate(cands) == ("alice", 10, 3)

cands2 = [("x", 1, 9)]
assert best_candidate(cands2) == ("x", 1, 9)

try:
    best_candidate([])
    assert False, "Expected ValueError on empty input"
except:
    pass
```

## **Exo 5**

```python
def comprehension_pack(nums: list[int]) -> tuple[list[int], set[int], dict[int, int]]:
    evens_squared = [x * x for x in nums if x % 2 == 0]
    abs_set = {abs(x) for x in nums}
    freq = {x: nums.count(x) for x in set(nums)}
    return evens_squared, abs_set, freq

# tests
nums = [2,-1,2,3,-1,4]
ev2, ab, fr = comprehension_pack(nums)
assert ev2 == [4,4,16]
assert ab == {1,2,3,4}
assert fr == {2: 2, -1: 2, 3: 1, 4: 1}
assert nums == [2, -1, 2, 3, -1, 4]
```
