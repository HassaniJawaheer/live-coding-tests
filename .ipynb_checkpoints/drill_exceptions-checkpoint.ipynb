## **Exo 1**

```python
def safe_ints(items: list[object]) -> list[int | None]:
    out: list[int | None] = []
    for x in items:
        try:
            out.append(int(x))
        except (ValueError, TypeError):
            out.append(None)
    return out

# Tests
assert safe_ints(["10", 3.2, None, "x", True]) == [10, 3, None, None, 1]
assert safe_ints([]) == []
assert safe_ints(["-7", " 8"]) == [-7, 8]
```
