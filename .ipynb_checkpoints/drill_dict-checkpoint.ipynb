## **Exo 1**

```python
def safe_lookup(d: dict[str, int], keys: list[str], default: int) -> list[int]:
    return [d.get(k, default) for k in keys]

d = {"a": 1, "b": 2}
assert safe_lookup(d, ["a", "c", "b", "c"], default=-1) == [1, -1, 2, -1]
assert d == {"a": 1, "b": 2}

assert safe_lookup({}, ["x"], default=0) == [0]
assert safe_lookup({"x": 7}, [], default=0) == []

d2 = {"k": 0}
assert safe_lookup(d2, ["k", "missing"], default=99) == [0, 99]
```

## **Exo 2**

```python
def histogram(words: list[str]) -> dict[str,int]:
    out = {}
    for word in words:
        out[word] = out.get(word, 0) + 1
    return out
# tests
assert histogram(["a", "b", "a"]) == {"a": 2, "b": 1}
assert histogram([]) == {}
assert histogram(["A", "a", "A", "b"]) == {"A": 2, "a": 1, "b": 1}
assert histogram(["x"]) == {"x": 1}
```

## **Exo 3**

```python
def group_by_first_letter(words: list[str]) -> dict[str, list[str]]:
    out: dict[str, list[str]] = {}
    for w in words:
        if w == "":
            continue
        out.setdefault(w[0], []).append(w)
    return out

# Tests
assert group_by_first_letter(["ab", "aX", "b1", "ba"]) == {"a": ["ab", "aX"], "b": ["b1", "ba"]}
assert group_by_first_letter(["", "a", "", "ab"]) == {"a": ["a", "ab"]}
assert group_by_first_letter([]) == {}
assert group_by_first_letter(["z"]) == {"z": ["z"]}
```

## **Exo 4**

```python
def merge_sums(left: dict[str, int], right: dict[str, int]) -> dict[str, int]:
    out = left.copy()       # copy left (no mutation)
    out.update(right)       # bring in right values (overwrites conflicts)
    for k, v in left.items():
        if k in right:
            out[k] = v + right[k]
    return out

# Tests
a = {"x": 1, "y": 2}
b = {"y": 5, "z": 9}
c = merge_sums(a,b)

assert c == {"x":1, "y": 7, "z": 9}
assert a == {"x": 1, "y": 2}
assert b == {"y": 5, "z": 9}

assert merge_sums({}, {"a": 1}) == {"a": 1}
assert merge_sums({"a": 1}, {}) == {"a":1}

d = merge_sums({"k": 0}, {"k": 0, "m": 2})
assert d == {"k": 0, "m": 2}
```
