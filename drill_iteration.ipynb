## **Exo 1**

```python
def find_mismatches(a: list[int], b: list[int]) -> list[int]:
    n = min(len(a), len(b))
    return [i for i, x in enumerate(a[:n]) if x != b[i]]

# Tests
assert find_mismatches([1, 2, 3], [1, 9, 3]) == [1]
assert find_mismatches([1, 2, 3], [1, 2]) == []
assert find_mismatches([], [1, 2]) == []
assert find_mismatches([0, 0, 1], [0, 1, 1]) == [1]
```

## **Exo 2**

```python
def pair_sums(a: list[int], b: list[int]) -> list[int]:
    return [x + y for x, y in zip(a, b)]

# Tests
assert pair_sums([1, 2, 3], [10, 20, 30]) == [11,22,33]
assert pair_sums([1, 2, 3], [10]) == [11]
assert pair_sums([], [1, 2]) == []
assert pair_sums([0, -1], [1, 1, 1]) == [1, 0]
```
