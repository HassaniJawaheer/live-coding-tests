## **Exo 2**

```python
def sum_optional(d: dict[str, int], keys: list[str]) -> int:
    out = 0
    for key in keys:
        try:
            out += d.get(key,0)
        except KeyError:
            pass
    return out

# Tests
d = {"a" : 1, "b": 2}
assert sum_optional(d, ["a", "c", "b", "c"]) == 3
assert sum_optional(d, []) == 0
assert sum_optional({}, ["x", "x"]) == 0
assert sum_optional({"x": 5}, ["x", "x"]) == 10
```
