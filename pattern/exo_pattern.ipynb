```python
def longest_unique_sequence(values: list[int]) -> tuple[int, list[int]]:
    if not values:
        return 0, []

    seen = set()
    left = 0

    best_start = 0
    best_length = 0

    for right, value in enumerate(values):
        while value in seen:
            seen.remove(values[left])
            left += 1

        seen.add(value)

        current_length = right - left + 1

        if current_length > best_length:
            best_length = current_length
            best_start = left

    return best_length, values[best_start:best_start + best_length]
assert longest_unique_sequence([]) == (0, [])

assert longest_unique_sequence([5]) == (1, [5])

assert longest_unique_sequence([1, 2, 3]) == (3, [1, 2, 3])

assert longest_unique_sequence([1, 1, 1]) == (1, [1])

assert longest_unique_sequence(
    [1, 2, 3, 2, 4, 5]
) == (4, [3, 2, 4, 5])
```
