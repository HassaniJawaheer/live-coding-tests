## Exercice 1 — Premier événement répété

```python
def first_repeated_event(events: list[str]) -> str | None:
    # Initialization
    seen_events = set()

    # Iterate over events
    for event in events:
        if event in seen_events:
            return event
        seen_events.add(event)
    return None

# Tests
assert first_repeated_event([]) is None

assert first_repeated_event(["a"]) is None

assert first_repeated_event(["a", "b", "c"]) is None

assert first_repeated_event(["a", "b", "a"]) == "a"

assert first_repeated_event(["a", "b", "b", "a"]) == "b"

assert first_repeated_event(["x", "x", "y"]) == "x"

assert first_repeated_event(["A", "a", "A"]) == "A"
```

## Exercice 2 — Détection de doublons après normalisation

```python
def find_duplicate_sensor(sensor_codes: list[str]) -> str | None:
    # Initialization
    seen_codes = set()

    # Iterate over codes
    for code in sensor_codes:
        normalized_code = code.strip().lower()
        if normalized_code in seen_codes:
            return code
        seen_codes.add(normalized_code)
    return None

# Tests
assert find_duplicate_sensor([]) is None

assert find_duplicate_sensor(
    ["TEMP_01", "PRESSURE_02", "HUM_03"]
) is None

assert find_duplicate_sensor(
    ["TEMP_01", "PRESSURE_02", "temp_01"]
) == "temp_01"

assert find_duplicate_sensor(
    ["  TEMP_01", "PRESSURE_02", "temp_01  "]
) == "temp_01  "

assert find_duplicate_sensor(
    ["A", " b ", "B", "a"]
) == "B"

assert find_duplicate_sensor(
    [" X ", "x", "X"]
) == "x"
```

## Exercice 3 — Nettoyage et déduplication d’événements

```python
def clean_unique_events(events: list[dict]) -> list[dict]:
    clean_events = []
    seen_id_events = set()
    required_keys = {"id", "type", "source"}

    for event in events:
        # Check required keys
        if not required_keys.issubset(event):
            continue

        # Check required values
        if not all(isinstance(event[key], str) for key in required_keys):
            continue

        normalized_id = event["id"].strip().lower()

        # Skip duplicate IDs
        if normalized_id in seen_id_events:
            continue

        seen_id_events.add(normalized_id)

        clean_events.append(
            {
                "id": normalized_id,
                "type": event["type"].strip().upper(),
                "source": event["source"].strip().lower(),
            }
        )

    return clean_events
# Tests
assert clean_unique_events([]) == []

assert clean_unique_events([
    {"id": " EVT-001 ", "type": " click ", "source": "WEB"},
]) == [
    {"id": "evt-001", "type": "CLICK", "source": "web"},
]

assert clean_unique_events([
    {"id": "A", "type": "click", "source": "web"},
    {"id": " a ", "type": "view", "source": "mobile"},
    {"id": "B", "type": " purchase ", "source": "API"},
]) == [
    {"id": "a", "type": "CLICK", "source": "web"},
    {"id": "b", "type": "PURCHASE", "source": "api"},
]

assert clean_unique_events([
    {"id": "A", "type": "click"},                     # source absent
    {"id": "B", "type": 42, "source": "web"},        # type invalide
    {"id": "C", "type": "view", "source": "mobile"},
]) == [
    {"id": "c", "type": "VIEW", "source": "mobile"},
]

# Vérifie que la première occurrence valide gagne.
assert clean_unique_events([
    {"id": "X", "type": 123, "source": "web"},       # invalide
    {"id": " x ", "type": "click", "source": "api"}, # première valide
    {"id": "X", "type": "view", "source": "mobile"}, # doublon
]) == [
    {"id": "x", "type": "CLICK", "source": "api"},
]

# Vérifie que les entrées ne sont pas modifiées.
original = [
    {"id": " A ", "type": " click ", "source": "WEB"},
]
clean_unique_events(original)

assert original == [
    {"id": " A ", "type": " click ", "source": "WEB"},
]
```
