## Exercice 5 — Compter les statuts de prédiction

```python
def count_prediction_statuses(statuses: list[str]) -> dict[str, int]:
    # Count the occurences of each statuses
    result = {}
    for status in statuses:
        result[status] = result.get(status, 0) + 1
    return result
# Tests
assert count_prediction_statuses([]) == {}

assert count_prediction_statuses(["success"]) == {
    "success": 1
}

assert count_prediction_statuses([
    "success", "error", "success", "timeout", "error", "success"
]) == {
    "success": 3,
    "error": 2,
    "timeout": 1,
}

assert count_prediction_statuses([
    "error", "error", "error"
]) == {
    "error": 3,
}

assert count_prediction_statuses([
    "error", "Error", "ERROR", "error"
]) == {
    "error": 2,
    "Error": 1,
    "ERROR": 1,
}
```

```python
from collections import Counter


def most_frequent_error(errors: list[str]) -> tuple[str, int] | None:
    if not errors:
        return None

    counts = Counter(errors)

    max_error = errors[0]
    max_occ = counts[max_error]

    for error, count in counts.items():
        if count > max_occ:
            max_error = error
            max_occ = count

    return max_error, max_occ

# Tests
assert most_frequent_error([]) is None

assert most_frequent_error(
    ["timeout"]
) == ("timeout", 1)

assert most_frequent_error(
    ["timeout", "network", "timeout", "auth", "network", "timeout"]
) == ("timeout", 3)

assert most_frequent_error(
    ["network", "timeout", "timeout", "network"]
) == ("network", 2)

assert most_frequent_error(
    ["a", "b", "c", "b", "c", "a"]
) == ("a", 2)

assert most_frequent_error(
    ["x", "y", "y", "x", "z", "y"]
) == ("y", 3)
```

## Exercice 6 — Détecter les événements anormalement fréquents

```python
from collections import Counter

def frequent_events(events: list[str], threshold: int) -> list[tuple[str, int]]:
    result = []

    counts = Counter(events)
    for event, count_event in counts.items():
        if count_event >= threshold:
            result.append(
                (event, count_event)
            )
    return result

# Tests
assert frequent_events([], 2) == []

assert frequent_events(
    ["a", "b", "a", "c", "b", "a"],
    2,
) == [
    ("a", 3),
    ("b", 2),
]

assert frequent_events(
    ["x", "y", "z"],
    1,
) == [
    ("x", 1),
    ("y", 1),
    ("z", 1),
]

assert frequent_events(
    ["a", "b", "a", "b", "c", "c"],
    2,
) == [
    ("a", 2),
    ("b", 2),
    ("c", 2),
]

assert frequent_events(
    ["error", "error", "ok", "error", "ok"],
    3,
) == [
    ("error", 3),
]

assert frequent_events(
    ["a", "a", "b"],
    3,
) == []
```

## Exercice 7 — Fréquence des mots normalisés

```python
from collections import Counter


def frequent_words(
    messages: list[str],
    min_count: int
) -> list[tuple[str, int]]:
    # Assemble all messages
    assembled_messages = " ".join(messages)

    # Split into individual words and normalize
    normalized_words = [
        word.lower()
        for word in assembled_messages.split()
    ]

    # Count occurrences
    word_counts = Counter(normalized_words)

    # Keep sufficiently frequent words
    result = []
    for word, count in word_counts.items():
        if count >= min_count:
            result.append((word, count))

    return result

# Tests
assert frequent_words([], 2) == []

assert frequent_words(
    ["Python python PYTHON"],
    2,
) == [
    ("python", 3),
]

assert frequent_words(
    [
        "Python is great",
        "python is useful",
        "PYTHON is great",
    ],
    2,
) == [
    ("python", 3),
    ("is", 3),
    ("great", 2),
]

assert frequent_words(
    [
        "error timeout",
        "ERROR network",
        "timeout error",
    ],
    2,
) == [
    ("error", 3),
    ("timeout", 2),
]

assert frequent_words(
    ["a b c", "d e f"],
    2,
) == []

assert frequent_words(
    ["  alpha   beta  ", "ALPHA beta gamma"],
    2,
) == [
    ("alpha", 2),
    ("beta", 2),
]
```
